```python
%sql
-- 1. Driver Dimension
CREATE OR REPLACE TABLE f1_project.default.dim_driver AS
SELECT DISTINCT
    driver_id,
    driver_code,
    driver_number,
    driver_name,
    driver_nationality
FROM f1_project.default.f1_project_data_silver;

-- 2. Constructor Dimension
CREATE OR REPLACE TABLE f1_project.default.dim_constructor AS
SELECT DISTINCT
    constructor_id,
    constructor_name
FROM f1_project.default.f1_project_data_silver;

-- 3. Circuit Dimension
CREATE OR REPLACE TABLE f1_project.default.dim_circuit AS
SELECT DISTINCT
    circuit_id,
    circuit_name
FROM f1_project.default.f1_project_data_silver;

-- 4. Race Results Fact Table
CREATE OR REPLACE TABLE f1_project.default.fact_race_results AS
SELECT
    season,
    round_number,
    race_date,
    circuit_id,
    driver_id,
    constructor_id,
    grid_position,
    position AS finishing_position,
    points,
    laps_completed,
    status,
    fastest_lap_number,
    fastest_lap_time
FROM f1_project.default.f1_project_data_silver;
```
